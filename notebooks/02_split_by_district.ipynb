```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv('../dataset/processed/sri_lanka_land_prices_cleaned.csv')
df.head()
```

```python
districts = df['District'].unique()
print(districts)
```

```python
district_folder = "../dataset/raw/districts"

os.makedirs(district_folder, exist_ok=True)
```

```python
for district in districts:

    district_df = df[df["District"] == district]

    filename = f"{district}.csv"

    filepath = os.path.join(
        district_folder,
        filename
    )

    district_df.to_csv(
        filepath,
        index=False
    )

    print(
        f"{district}: {len(district_df)} rows saved"
    )
```
