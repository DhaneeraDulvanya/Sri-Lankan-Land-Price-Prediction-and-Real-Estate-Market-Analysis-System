```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../dataset/processed/cleaned_house_prices.csv')
df.head()
```

```python
target = "PRICE"
```

```python
columns_to_drop = [
    "STREET_NAME",
    "BROKERTITLE",
    "MAIN_ADDRESS",
    "FORMATTED_ADDRESS",
    "ADMINISTRATIVE_AREA_LEVEL_2"
]

df_fe = df.drop(
    columns=columns_to_drop,
    errors="ignore"
)

print(df_fe.columns)
```

```python
df_fe.isnull().sum()
```

```python
categorical_features = [
    "TYPE",
    "LOCALITY",
    "ADDRESS",
    "STATE"
]

numerical_features = [
    "PRICE",
    "PROPERTYSQFT",
    "BEDS",
    "BATH"
]
```

```python
# price per square foot
df_fe["PRICE_PER_SQFT"] = df_fe["PRICE"] / df_fe["PROPERTYSQFT"]
```

```python
# value category

def size_category(sqft):

    if sqft < 1000:
        return "Small"

    elif sqft < 2000:
        return "Medium"

    elif sqft < 3500:
        return "Large"

    else:
        return "Very Large"


df_fe["SIZE_CATEGORY"] = (
    df_fe["PROPERTYSQFT"]
    .apply(size_category)
)
```

```python
df_fe["SIZE_CATEGORY"].value_counts()
```

```python
df_fe.head()
```

```python
# beads per bath ratio

df_fe["BEDS_PER_BATH"] = df_fe["BEDS"] / df_fe["BATH"].replace(0, np.nan)

df_fe["BEDS_PER_BATH"] = (
    df_fe["BEDS_PER_BATH"]
    .fillna(0)
)
```

```python
df_fe.head()
```

```python
# total room related feature

df_fe["TOTAL_BEDS_BATH"] = (
    df_fe["BEDS"] + df_fe["BATH"]
)
```

```python
features = [
    "BEDS",
    "BATH",
    "PROPERTYSQFT",
    "TYPE",
    "STATE",
    "LOCALITY"
]

X = df_fe[features]
y = df_fe[target]

print("X shape:", X.shape)
print("y shape:", y.shape)
```
