```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../dataset/raw/Colombo-House-Dataset-3000.csv')
df.head()
```

```python
print(df.columns)
```

```python
df.info()
```

```python
print(df["PRICE"].dtype)
print(df["PROPERTYSQFT"].dtype)
```

```python
df["PRICE"] = (
    df["PRICE"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["PROPERTYSQFT"] = (
    df["PROPERTYSQFT"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

print(df["PRICE"].dtype)
print(df["PROPERTYSQFT"].dtype)
```

```python
# check for missing values
missing_values = df.isnull().sum()

print(missing_values)
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
)

missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})

missing_table
```

```python
df.drop(columns=['STREET_NAME'], inplace=True)
```
