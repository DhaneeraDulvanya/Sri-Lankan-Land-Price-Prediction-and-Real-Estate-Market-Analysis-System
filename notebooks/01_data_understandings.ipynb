### Visualizations

```python
df.head()
```

```python
# price per perch destribution

plt.figure(figsize=(10, 6))
sns.histplot(df["Price_Per_Perch"], bins=1000, kde=True)

plt.title('Price per Perch Distribution')
plt.xlabel('Price per Perch')
plt.ylabel('Frequency')
plt.show()
```

```python
# land size vs total price
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Perches', y='Total_Price', data=df)

plt.title('Land Size vs Total Price')
plt.xlabel('Land Size (perches)')
plt.ylabel('Total Price (LKR)')
plt.show()
```

```python
# total price vs distance to the town
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Distance_To_Town_km', y='Total_Price', data=df)
plt.title('Total Price vs Distance to the Town')
plt.xlabel('Distance to the Town (km)')
plt.ylabel('Total Price (LKR)')
plt.show()
```

```python
# price per perch vs distance to the town
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Distance_To_Town_km', y='Price_Per_Perch', data=df)
plt.title('Price per Perch vs Distance to the Town')
plt.xlabel('Distance to the Town (km)')
plt.ylabel('Price per Perch (LKR)')
plt.show()
```

## Analyze the target variable

```python
df["Price_Per_Perch"].describe()
```

```python
ppp_mean = df["Price_Per_Perch"].mean()
ppp_median = df["Price_Per_Perch"].median()

print(f"Mean of Price per Perch: {ppp_mean}")
print(f"Median of Price per Perch: {ppp_median}")
```

```python
# check for outliers in Price_Per_Perch using boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(x=df["Price_Per_Perch"])

plt.title('Boxplot of Price per Perch')
plt.xlabel('Price per Perch (LKR)')
plt.ylabel('Frequency')
plt.show()
```

## District analysis

```python
district_price = (
    df.groupby('District')['Price_Per_Perch'].mean().sort_values(ascending=False)
)

print(district_price)  
```

```python
# visualize the average price per perch by district
plt.figure(figsize=(12, 6))
sns.barplot(x=district_price.index, y=district_price.values)
plt.title('Average Price per Perch by District')
plt.xlabel('District')
plt.ylabel('Average Price per Perch (LKR)')
plt.xticks(rotation=70)
plt.show()
```

```python
# meadian price per perch by district
dis_price = (
    df.groupby('District')['Price_Per_Perch'].median().sort_values(ascending=False)
)

print(dis_price)
```

```python
# visualize the median price per perch by district
plt.figure(figsize=(12, 6))
sns.barplot(x=dis_price.index, y=dis_price.values)
plt.title('Median Price per Perch by District')
plt.xlabel('District')
plt.ylabel('Median Price per Perch (LKR)')
plt.xticks(rotation=70)
plt.show()
```

```python
# perches vs price per perch
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Perches', y='Price_Per_Perch', data=df)
plt.title('Perches vs Price per Perch')
plt.xlabel('Land Size (perches)')
plt.ylabel('Price per Perch (LKR)')
plt.show()
```
