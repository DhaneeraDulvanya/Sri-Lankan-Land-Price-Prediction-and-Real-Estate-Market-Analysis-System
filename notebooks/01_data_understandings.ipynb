### Visualizations

```python
df.head()
```

```python
# price per perch destribution

plt.figure(figsize=(10, 6))
sns.histplot(df["Price_Per_Perch"], bins=1000, kde=True)

plt.title('Price per Perch Distribution')
plt.xlabel('Price per Perch')
plt.ylabel('Frequency')
plt.show()
```

```python
# land size vs total price
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Perches', y='Total_Price', data=df)

plt.title('Land Size vs Total Price')
plt.xlabel('Land Size (perches)')
plt.ylabel('Total Price (LKR)')
plt.show()
```

```python
# total price vs distance to the town
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Distance_To_Town_km', y='Total_Price', data=df)
plt.title('Total Price vs Distance to the Town')
plt.xlabel('Distance to the Town (km)')
plt.ylabel('Total Price (LKR)')
plt.show()
```
